```python
import numpy as np

# Create a numeric vector using your first name
Fname = np.array([1234, 12, 25, 35, 45, 55, 65, 75, 85, 95])

print("Original Data:")
print(Fname)


# 1(a). Normalize to range [-3, +3]
min_val = np.min(Fname)
max_val = np.max(Fname)

normalize_3 = -3 + ((Fname - min_val) / (max_val - min_val)) * 6

print("\nNormalization to [-3, +3]:")
print(normalize_3)


# 1(b). Normalize to range [-1, +1]
normalize_1 = -1 + ((Fname - min_val) / (max_val - min_val)) * 2

print("\nNormalization to [-1, +1]:")
print(normalize_1)


# 1(c). Min-Max Normalization to [0, 1]
min_max = (Fname - min_val) / (max_val - min_val)

print("\nMin-Max Normalization [0, 1]:")
print(min_max)


# 2. Z-Score Normalization
mean_val = np.mean(Fname)
std_val = np.std(Fname)

z_score = (Fname - mean_val) / std_val

print("\nZ-Score Normalization:")
print(z_score)


# 3. Decimal Scaling Normalization
max_abs = np.max(np.abs(Fname))

j = len(str(int(max_abs)))

decimal_scaling = Fname / (10 ** j)

print("\nDecimal Scaling Normalization:")
print(decimal_scaling)
```

```python
import pandas as pd
import numpy as np

# 1. Import the dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/diabetes.csv"
diabetes = pd.read_csv(url)

# 2. Display the first 10 records
print("2. First 10 Records:")
print(diabetes.head(10))

# 3. Display the last 10 records
print("\n3. Last 10 Records:")
print(diabetes.tail(10))

# 4. Total number of rows and columns
print("\n4. Number of Rows and Columns:")
print(diabetes.shape)

# 5. Display names of all variables
print("\n5. Column Names:")
print(diabetes.columns.tolist())

# 6. Display data types of each variable
print("\n6. Data Types:")
print(diabetes.dtypes)

# 7. Display summary statistics
print("\n7. Summary Statistics:")
print(diabetes.describe(include="all"))

# 8. Number of missing values in each column
print("\n8. Missing Values in Each Column:")
print(diabetes.isnull().sum())

# 9. Total number of missing values
print("\n9. Total Missing Values:")
print(diabetes.isnull().sum().sum())

# 10. Number of missing values in each column
print("\n10. Missing Values in Each Column:")
print(diabetes.isna().sum())

# 11. Identify rows containing at least one missing value
print("\n11. Rows with Missing Values:")
print(diabetes[diabetes.isnull().any(axis=1)])

# 12. Remove all rows containing missing values
cleaned_diabetes = diabetes.dropna()
print("\n12. Cleaned Dataset:")
print(cleaned_diabetes)

# 13. Replace missing values in a numerical column using mean
# Example: Glucose
if "Glucose" in diabetes.columns:
    diabetes["Glucose"] = diabetes["Glucose"].fillna(
        diabetes["Glucose"].mean()
    )
    print("\n13. Missing Glucose Values Replaced with Mean")

# 14. Replace missing values in another numerical column using median
# Example: BloodPressure
if "BloodPressure" in diabetes.columns:
    diabetes["BloodPressure"] = diabetes["BloodPressure"].fillna(
        diabetes["BloodPressure"].median()
    )
    print("14. Missing BloodPressure Values Replaced with Median")

# 15. Replace missing values in a categorical column using mode
# Example: Outcome, if treated as a categorical column
if "Outcome" in diabetes.columns:
    mode_value = diabetes["Outcome"].mode()
    if not mode_value.empty:
        diabetes["Outcome"] = diabetes["Outcome"].fillna(mode_value[0])
    print("15. Missing Outcome Values Replaced with Mode")

# 16. Forward filling: replace missing values with previous value
forward_filled = diabetes.ffill()
print("\n16. Forward Filled Dataset:")
print(forward_filled)

# 17. Backward filling: replace missing values with next value
backward_filled = diabetes.bfill()
print("\n17. Backward Filled Dataset:")
print(backward_filled)

# 18. Replace remaining missing values with a constant
constant_filled = diabetes.fillna(0)
print("\n18. Missing Values Replaced with 0:")
print(constant_filled)

# 19. Remove duplicate records
diabetes = diabetes.drop_duplicates()
print("\n19. Dataset After Removing Duplicates:")
print(diabetes)

# 20. Convert appropriate variables into categorical data type
if "Outcome" in diabetes.columns:
    diabetes["Outcome"] = diabetes["Outcome"].astype("category")
print("\n20. Data Types After Conversion:")
print(diabetes.dtypes)

# 21. Rename one or more column names
diabetes = diabetes.rename(
    columns={
        "BloodPressure": "Blood_Pressure",
        "SkinThickness": "Skin_Thickness"
    }
)
print("\n21. Renamed Columns:")
print(diabetes.columns.tolist())

# 22. Create a new derived column using existing variables
if "Glucose" in diabetes.columns and "Insulin" in diabetes.columns:
    diabetes["Glucose_Insulin_Sum"] = (
        diabetes["Glucose"] + diabetes["Insulin"]
    )
print("\n22. Dataset with New Derived Column:")
print(diabetes.head())

# 23. Display summary statistics
print("\n23. Summary Statistics:")
print(diabetes.describe(include="all"))

# 24. Find minimum and maximum values
numeric_columns = diabetes.select_dtypes(include=np.number).columns
print("\n24. Minimum Values:")
print(diabetes[numeric_columns].min())

print("\nMaximum Values:")
print(diabetes[numeric_columns].max())

# 25. Calculate range of a numerical variable
# Example: Glucose
if "Glucose" in diabetes.columns:
    glucose_range = (
        diabetes["Glucose"].max() - diabetes["Glucose"].min()
    )
    print("\n25. Range of Glucose:")
    print(glucose_range)

# 26. Calculate variance
print("\n26. Variance:")
print(diabetes[numeric_columns].var())

# 27. Calculate standard deviation
print("\n27. Standard Deviation:")
print(diabetes[numeric_columns].std())

# 28. Calculate Interquartile Range (IQR)
print("\n28. Interquartile Range (IQR):")
iqr = (
    diabetes[numeric_columns].quantile(0.75)
    - diabetes[numeric_columns].quantile(0.25)
)
print(iqr)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Import the dataset using the given link
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/diabetes.csv"
df = pd.read_csv(url)

# 2. Display first and last six records
print("First 6 records:")
print(df.head(6))

print("\nLast 6 records:")
print(df.tail(6))

# 3. Total number of missing values
print("\nTotal missing values:")
print(df.isnull().sum().sum())

# 4. Display rows containing missing values
print("\nRows with missing values:")
print(df[df.isnull().any(axis=1)])

# 5. Remove columns having more than 70% missing values
df = df.loc[:, df.isnull().mean() <= 0.70]
print("\nAfter removing columns with >70% missing values:")
print(df.head())

# Identify numerical and categorical columns
num_cols = df.select_dtypes(include=np.number).columns.tolist()
cat_cols = df.select_dtypes(include=["object", "category"]).columns.tolist()

# 6. Replace missing numerical values with mean
if len(num_cols) > 0:
    df[num_cols[0]] = df[num_cols[0]].fillna(df[num_cols[0]].mean())

# 7. Replace missing numerical values with median
if len(num_cols) > 1:
    df[num_cols[1]] = df[num_cols[1]].fillna(df[num_cols[1]].median())

# 8. Replace missing categorical values with mode
if len(cat_cols) > 0:
    mode_value = df[cat_cols[0]].mode()
    if not mode_value.empty:
        df[cat_cols[0]] = df[cat_cols[0]].fillna(mode_value[0])

# 9. Identify duplicate rows
print("\nDuplicate rows:")
print(df[df.duplicated()])

# 10. Remove duplicate rows
df = df.drop_duplicates()

# 11. Display data types
print("\nData types:")
print(df.dtypes)

# 12. Convert character columns into categorical variables
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].astype("category")

print("\nUpdated data types:")
print(df.dtypes)

# 13. Convert text values to lowercase
for col in df.select_dtypes(include=["object", "category"]).columns:
    df[col] = df[col].astype("string").str.lower()

print("\nText converted to lowercase:")
print(df.head())

# Refresh numerical columns
num_cols = df.select_dtypes(include=np.number).columns.tolist()

# 14. Detect outliers using boxplots
for col in num_cols:
    plt.figure(figsize=(6, 3))
    sns.boxplot(x=df[col])
    plt.title("Boxplot of " + col)
    plt.show()

# 15. Identify outliers using IQR
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    outliers = df[
        (df[col] < Q1 - 1.5 * IQR) |
        (df[col] > Q3 + 1.5 * IQR)
    ]

    print("\nOutliers in", col, ":")
    print(outliers[[col]])

# 16. Remove extreme outliers using 3 * IQR
df_clean = df.copy()

for col in num_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    df_clean = df_clean[
        (df_clean[col] >= Q1 - 3 * IQR) &
        (df_clean[col] <= Q3 + 3 * IQR)
    ]

print("\nDataset after removing extreme outliers:")
print(df_clean)

# 17. Create a histogram
if len(num_cols) > 0:
    col = num_cols[0]

    plt.figure(figsize=(6, 4))
    plt.hist(df_clean[col].dropna(), bins=15, edgecolor="black")
    plt.title("Histogram of " + col)
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()

# 18. Histogram with density curve
    plt.figure(figsize=(6, 4))
    sns.histplot(data=df_clean, x=col, kde=True)
    plt.title("Histogram with Density Curve")
    plt.show()

# 19. Violin plot
    plt.figure(figsize=(6, 4))
    sns.violinplot(data=df_clean, y=col)
    plt.title("Violin Plot")
    plt.show()

# 20. Dot plot for a small numerical dataset
values = [10, 12, 12, 15, 17, 17, 20, 22, 25]

plt.figure(figsize=(6, 3))
plt.plot(values, np.zeros(len(values)), "o")
plt.yticks([])
plt.xlabel("Values")
plt.title("Dot Plot")
plt.show()

# 21. Frequency table for Outcome
if "Outcome" in df_clean.columns:
    cat_col = "Outcome"
else:
    cat_cols = df_clean.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()

    if cat_cols:
        cat_col = cat_cols[0]
    else:
        cat_col = None

if cat_col is not None:
    freq = df_clean[cat_col].value_counts()

    print("\nFrequency table:")
    print(freq)

    # 22. Proportion and percentage
    print("\nProportion:")
    print(df_clean[cat_col].value_counts(normalize=True))

    print("\nPercentage:")
    print(df_clean[cat_col].value_counts(normalize=True) * 100)

    # 23. Bar chart
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df_clean, x=cat_col)
    plt.title("Bar Chart")
    plt.show()

    # 24. Horizontal bar chart
    freq.sort_values().plot(kind="barh")
    plt.title("Horizontal Bar Chart")
    plt.show()

    # 25. Percentage bar chart
    (freq / freq.sum() * 100).plot(kind="bar")
    plt.ylabel("Percentage")
    plt.title("Percentage Bar Chart")
    plt.show()

    # 26. Pie chart
    freq.plot(kind="pie", autopct="%1.1f%%")
    plt.title("Pie Chart")
    plt.ylabel("")
    plt.show()

    # 27. Lollipop chart
    f = freq.sort_values()

    plt.figure(figsize=(6, 4))
    plt.hlines(y=f.index.astype(str), xmin=0, xmax=f.values)
    plt.plot(f.values, f.index.astype(str), "o")
    plt.title("Lollipop Chart")
    plt.xlabel("Frequency")
    plt.show()

    # 28. Dot chart
    plt.figure(figsize=(6, 4))
    plt.plot(freq.values, freq.index.astype(str), "o")
    plt.title("Dot Chart")
    plt.xlabel("Frequency")
    plt.show()

# 29. Density plot
if len(num_cols) > 0:
    plt.figure(figsize=(6, 4))
    sns.kdeplot(data=df_clean, x=num_cols[0], fill=True)
    plt.title("Density Plot")
    plt.show()

# 30. Facet plot
if len(num_cols) > 0 and cat_col is not None:
    plot_data = df_clean.dropna(subset=[num_cols[0], cat_col])

    g = sns.FacetGrid(plot_data, col=cat_col, height=3)
    g.map_dataframe(sns.histplot, x=num_cols[0], bins=15)
    g.set_axis_labels(num_cols[0], "Frequency")
    g.set_titles("{col_name}")
    plt.show()

print("\nFinal cleaned dataset:")
print(df_clean.head())

print("\nFinal dataset shape:")
print(df_clean.shape)
```

```python
# AirPassengers Dataset - Time Series Analysis

import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.arima.model import ARIMA

# 1. Load the built-in AirPassengers dataset
# Python does not include the R dataset directly, so create it using
# the monthly passenger values from the classic AirPassengers dataset.

passengers = [
    112,118,132,129,121,135,148,148,136,119,104,118,
    115,126,141,135,125,149,170,170,158,133,114,140,
    145,150,178,163,172,178,199,199,184,162,146,166,
    171,180,193,181,183,218,230,242,209,191,172,194,
    196,196,236,235,229,243,264,272,237,211,180,201,
    204,188,235,227,234,264,302,293,259,229,203,229,
    242,233,267,269,270,315,364,347,312,274,237,278,
    284,277,317,313,318,374,413,405,355,306,271,306,
    315,301,356,348,355,422,465,467,404,347,305,336,
    340,318,362,348,363,435,491,505,404,359,310,337,
    360,342,406,396,420,472,548,559,463,407,362,405,
    417,391,419,461,472,535,622,606,508,461,390,432
]

# Create monthly dates from January 1949
dates = pd.date_range(start="1949-01-01", periods=len(passengers), freq="MS")

AirPassengers = pd.Series(
    passengers,
    index=dates,
    name="Passengers"
)

# 2. Explore the first five and last five observations
print("First five observations:")
print(AirPassengers.head())

print("\nLast five observations:")
print(AirPassengers.tail())


# 3. Identify the starting and ending time periods
print("\nStarting period:", AirPassengers.index[0])
print("Ending period:", AirPassengers.index[-1])


# 4. Determine frequency and time-series characteristics
print("\nFrequency: 12 observations per year (monthly)")
print("Data type:", type(AirPassengers))
print("Number of observations:", len(AirPassengers))
print("Time-series characteristics:")
print(AirPassengers.index.freq)


# 5. Visualize monthly passenger trend
plt.figure(figsize=(12, 5))
plt.plot(AirPassengers, label="Monthly Passengers")
plt.title("Monthly Air Passengers (1949–1960)")
plt.xlabel("Year")
plt.ylabel("Number of Passengers (Thousands)")
plt.legend()
plt.grid(True)
plt.show()


# 6. Decompose the time series
decomposition = seasonal_decompose(
    AirPassengers,
    model="multiplicative",
    period=12
)


# 7. Plot trend, seasonal, and random components
decomposition.plot()
plt.suptitle("AirPassengers Time Series Decomposition", y=1.02)
plt.show()


# 8. Analyze ACF and PACF
plt.figure(figsize=(12, 4))
plot_acf(AirPassengers, lags=36)
plt.title("Autocorrelation Function (ACF)")
plt.show()

plt.figure(figsize=(12, 4))
plot_pacf(AirPassengers, lags=36, method="ywm")
plt.title("Partial Autocorrelation Function (PACF)")
plt.show()


# 9. Fit an ARIMA model
# ARIMA(2, 1, 2) is used as an example.
model = ARIMA(
    AirPassengers,
    order=(2, 1, 2)
)

fitted_model = model.fit()


# 10. Display model parameters and summary
print("\nARIMA Model Parameters:")
print(fitted_model.params)

print("\nARIMA Model Summary:")
print(fitted_model.summary())


# 11. Forecast the next 12 months
forecast_result = fitted_model.get_forecast(steps=12)

forecast = forecast_result.predicted_mean
confidence_intervals = forecast_result.conf_int()

print("\nForecast for the next 12 months:")
print(forecast)


# 12. Plot actual and forecasted passenger values
plt.figure(figsize=(12, 5))

plt.plot(
    AirPassengers,
    label="Actual Passengers"
)

plt.plot(
    forecast,
    color="red",
    label="Forecasted Passengers"
)

plt.fill_between(
    confidence_intervals.index,
    confidence_intervals.iloc[:, 0],
    confidence_intervals.iloc[:, 1],
    color="pink",
    alpha=0.3,
    label="95% Confidence Interval"
)

plt.title("AirPassengers: Actual vs Forecasted Values")
plt.xlabel("Year")
plt.ylabel("Number of Passengers (Thousands)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Assessment 8: K-Means Clustering

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# 1. Load the dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/diabetes.csv"

df = pd.read_csv(url)

# 2. View the dataset
print("First five rows:")
print(df.head())

print("\nDataset information:")
print(df.info())

print("\nDataset dimensions:")
print(df.shape)

# 3. Check missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

# Remove rows containing missing values
df = df.dropna()

# 4. Check dependent variable distribution
# Outcome is the target variable in this dataset
if "Outcome" in df.columns:
    print("\nOutcome distribution:")
    print(df["Outcome"].value_counts())

# 5. Select only numerical features
# Exclude Outcome because it is the target variable
numeric_df = df.select_dtypes(include=np.number)

if "Outcome" in numeric_df.columns:
    X = numeric_df.drop(columns=["Outcome"])
else:
    X = numeric_df.copy()

print("\nNumerical features selected:")
print(X.head())

# 6. Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("\nStandardized data:")
print(X_scaled[:5])

# 7. Apply K-Means with K = 2
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

kmeans.fit(X_scaled)

# 8. Display K-Means results
print("\nK-Means Results:")
print("Cluster assignments:")
print(kmeans.labels_)

print("\nWithin-cluster sum of squares (inertia):")
print(kmeans.inertia_)

# Add cluster labels to the original data
df["Cluster_K2"] = kmeans.labels_

print("\nDataset with cluster assignments:")
print(df.head())

# 9. Display cluster centers
centers = pd.DataFrame(
    scaler.inverse_transform(kmeans.cluster_centers_),
    columns=X.columns
)

print("\nCluster Centers (original scale):")
print(centers)

# 10. Calculate WCSS for K = 1 to 10
wcss = []

for k in range(1, 11):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)
    wcss.append(model.inertia_)

print("\nWCSS values:")
for k, value in enumerate(wcss, start=1):
    print("K =", k, "WCSS =", round(value, 2))

# 11. Display the elbow plot
plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    wcss,
    marker="o"
)

plt.title("Elbow Method for Optimal K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()

# 12. Apply K-Means based on the elbow recommendation
# Inspect the elbow plot and set the selected K below.
# Example: K = 3. Change this if the elbow suggests another value.

optimal_k = 3

final_kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

df["Optimal_Cluster"] = final_kmeans.fit_predict(X_scaled)

print("\nCluster assignments for optimal K =", optimal_k)
print(df["Optimal_Cluster"].value_counts().sort_index())

print("\nFinal cluster centers (original scale):")
final_centers = pd.DataFrame(
    scaler.inverse_transform(final_kmeans.cluster_centers_),
    columns=X.columns
)

print(final_centers)
```

# Part B: Hierarchical Clustering

```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.spatial.distance import pdist

# 1. Load the dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/diabetes.csv"

df_h = pd.read_csv(url)

# 2. View the dataset
print("First five rows:")
print(df_h.head())

print("\nDataset dimensions:")
print(df_h.shape)

# 3. Check missing values
print("\nMissing values:")
print(df_h.isnull().sum())

# Remove rows containing missing values
df_h = df_h.dropna()

# 4. Select only numerical attributes
numeric_df = df_h.select_dtypes(include=np.number)

# Exclude Outcome if it exists
if "Outcome" in numeric_df.columns:
    X_h = numeric_df.drop(columns=["Outcome"])
else:
    X_h = numeric_df.copy()

print("\nSelected numerical attributes:")
print(X_h.head())

# 5. Standardize the data
scaler_h = StandardScaler()
X_h_scaled = scaler_h.fit_transform(X_h)

print("\nStandardized data:")
print(X_h_scaled[:5])

# 6. Calculate Euclidean distance
distance_matrix = pdist(
    X_h_scaled,
    metric="euclidean"
)

print("\nFirst 10 Euclidean distances:")
print(distance_matrix[:10])

# 7. Perform hierarchical clustering using Ward's method
Z = linkage(
    X_h_scaled,
    method="ward",
    metric="euclidean"
)

print("\nHierarchical clustering linkage matrix:")
print(Z[:5])

# 8. Plot the dendrogram
plt.figure(figsize=(12, 6))

dendrogram(
    Z,
    truncate_mode="lastp",
    p=30
)

plt.title("Hierarchical Clustering Dendrogram (Ward's Method)")
plt.xlabel("Cluster or Sample")
plt.ylabel("Distance")
plt.show()

# 9. Divide the data into 3 clusters
clusters = fcluster(
    Z,
    t=3,
    criterion="maxclust"
)

df_h["Cluster"] = clusters

print("\nCluster assignments:")
print(df_h[["Cluster"]].head(20))

# 10. Display the number of samples in each cluster
print("\nNumber of samples in each cluster:")
print(df_h["Cluster"].value_counts().sort_index())
```

```python
# Principal Component Analysis (PCA)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Load the dataset and required libraries

url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/diabetes.csv"

data = pd.read_csv(url)

print("Dataset loaded successfully!")
print(data.head())


# 2. Display total number of rows and columns

print("\nNumber of rows:", data.shape[0])
print("Number of columns:", data.shape[1])


# 3. Extract numeric columns, clean the data, and normalize it

# Select only numerical columns
numeric_data = data.select_dtypes(include=np.number)

# Remove rows containing missing values
numeric_data = numeric_data.dropna()

# Remove columns with zero variance
numeric_data = numeric_data.loc[
    :, numeric_data.std() != 0
]

# Standardize the data
scaler = StandardScaler()

data_scaled = scaler.fit_transform(numeric_data)

print("\nStandardized data (first six rows):")
print(pd.DataFrame(
    data_scaled,
    columns=numeric_data.columns
).head())


# 4. Perform PCA and display the summary

pca = PCA()

pca_result = pca.fit_transform(data_scaled)

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("\nPercentage of Variance Explained:")
print(pca.explained_variance_ratio_ * 100)

print("\nCumulative Variance Explained:")
print(np.cumsum(pca.explained_variance_ratio_) * 100)


# 5. Scree plot with percentage labels and y-axis from 0 to 60

variance = pca.explained_variance_ratio_ * 100

components = [
    "PC" + str(i)
    for i in range(1, len(variance) + 1)
]

plt.figure(figsize=(9, 5))

bars = plt.bar(
    components,
    variance,
    color="skyblue",
    edgecolor="black"
)

for bar, value in zip(bars, variance):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.5,
        f"{value:.2f}%",
        ha="center"
    )

plt.ylim(0, 60)
plt.xlabel("Principal Components")
plt.ylabel("Percentage of Variance Explained")
plt.title("PCA Scree Plot")
plt.show()


# 6. Create PCA biplot with non-overlapping labels

# Use the first two principal components
scores = pca_result[:, :2]

# Loadings for the first two components
loadings = pca.components_[:2].T

plt.figure(figsize=(10, 7))

# Plot observations
plt.scatter(
    scores[:, 0],
    scores[:, 1],
    alpha=0.5,
    label="Observations"
)

# Scale loading arrows for visibility
scale = np.max(np.abs(scores)) * 0.7

for i, feature in enumerate(numeric_data.columns):
    x = loadings[i, 0] * scale
    y = loadings[i, 1] * scale

    plt.arrow(
        0, 0, x, y,
        color="red",
        head_width=0.08,
        length_includes_head=True
    )

    plt.text(
        x * 1.1,
        y * 1.1,
        feature,
        color="red",
        ha="center"
    )

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Biplot of Diabetes Dataset")
plt.axhline(0, color="gray", linewidth=0.5)
plt.axvline(0, color="gray", linewidth=0.5)
plt.grid(True)
plt.show()


# 7. Extract PCA scores into a DataFrame named pca_scores

pca_scores = pd.DataFrame(
    pca_result,
    columns=components
)

print("\nFirst six PCA scores:")
print(pca_scores.head(6))


# 8. Scatter plot of PC1 vs PC2 with blue circular points

registration_number = "23BDS0247"

plt.figure(figsize=(8, 5))

plt.scatter(
    pca_scores["PC1"],
    pca_scores["PC2"],
    color="blue",
    marker="o"
)

plt.xlabel("Principal Component 1 (PC1)")
plt.ylabel("Principal Component 2 (PC2)")

plt.title(
    "PCA Scatter Plot - registration_number: "
    + registration_number
)

plt.grid(True)
plt.show()
```
